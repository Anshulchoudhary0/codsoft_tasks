```python
def do_overlap(L1, R1, L2, R2):
    # L1, R1, L2, R2 are tuples or lists containing (x, y)

    # 1. If one rectangle is on the left side of the other
    if L1[0] > R2[0] or L2[0] > R1[0]:
        return 0

    # 2. If one rectangle is above the other
    if R1[1] > L2[1] or R2[1] > L1[1]:
        return 0

    # If none of the above are true, they overlap
    return 1

# --- Example Walkthrough ---
L1 = (0, 10)
R1 = (10, 0)
L2 = (5, 5)
R2 = (15, 0)

print(do_overlap(L1, R1, L2, R2))  # Output: 1
```
